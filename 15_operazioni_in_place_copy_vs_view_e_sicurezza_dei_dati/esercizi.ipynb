```python
""" 
Esercizio 1
Creare un dataframe con prodotti di un negozio, 
selezionare quelli con prezzo > 50 usando boolean indexing e creare copia indipendente. 
Aggiungere colonna “Costoso” (True se prezzo > 100) 
ordinare in-place per prezzo decrescente.

"""

import numpy as np
import pandas as pd

rng = np.random.default_rng(seed = 42)

n_prodotti = 100
prodotti_list = [f'Prodotto_{i:04d}' for i in range(1, n_prodotti + 1)]
n = len(prodotti_list)
prezzi_list = rng.uniform(10.0, 300.0, size=n)
categorie_prodotti = [
    "Elettronica e Tecnologia",
    "Abbigliamento e Moda",
    "Casa e Cucina",
    "Sport e Tempo Libero",
    "Salute e Benessere",
    "Libri e Cartoleria",
    "Giocattoli e Infanzia",
    "Alimentari e Bevande",
    "Giardinaggio e Fai da te",
    "Animali Domestici",
]

negozio_df = pd.DataFrame({
    "prodotti_id":prodotti_list,
    "prezzo_unitario":prezzi_list,
    "categoria": rng.choice(categorie_prodotti, size=n),
})

# selezionare quelli con prezzo > 50 usando boolean indexing e creare copia indipendente. 
negozio_subset_copy = negozio_df[negozio_df['prezzo_unitario']>50].copy()

print("LISTA DI PRODOTTI CON PREZZO > 50\n", negozio_subset_copy)

# Aggiungere colonna “Costoso” (True se prezzo > 100) ordinare in-place per prezzo decrescente.
negozio_df['costoso'] = negozio_df['prezzo_unitario']>100

# Ordinare in-place per prezzo decrescente
negozio_df.sort_values(by='prezzo_unitario', ascending=False, inplace=True)

print('\nPRODOTTI ORDINATE PER PREZZO DECRESCENTE\n', negozio_df.reset_index())

```

```python
""" 
Esercizio 2
Creare un dataframe con libri (titolo,autore,pagine,prezzo). 
Selezionare libri con più di 300 pagine e creare copia indipendente. 
Aggiungere colonna “Lungo” (True se pagine > 400). 
Ordinare per prezzo in ordine crescente in-place.

"""

import numpy as np 
import pandas as pd 

rng = np.random.default_rng(seed = 111)

# Creiamo una lista di dizionari con i dati già abbinati coerentemente
dati_libri = [
    {   
        "autore": "J.R.R. Tolkien", 
        "genere": "Fantasy", 
        "nazionalita": "Britannica"
    },
    {
        "autore": "Agatha Christie",
        "genere": "Giallo e Thriller",
        "nazionalita": "Britannica",
    },
    {
        "autore": "Isaac Asimov",
        "genere": "Fantascienza",
        "nazionalita": "Statunitense",
    },
    {
        "autore": "Italo Calvino",
        "genere": "Narrativa Contemporanea",
        "nazionalita": "Italiana",
    },
    {
        "autore": "Stephen King",
        "genere": "Giallo e Thriller",
        "nazionalita": "Statunitense",
    },
    {
        "autore": "Dante Alighieri",
        "genere": "Classici",
        "nazionalita": "Italiana",
    },
]

biblioteca_df = pd.DataFrame(dati_libri) # creo dataframe
biblioteca_df['titolo'] = libri_list = [ f'Libro_{i:04d}' for i in range(1, len(biblioteca_df) + 1)] # riempio la colonna titolo con la lista libri creata in modo casuale
biblioteca_df['pagine'] = rng.integers(50, 600, len(biblioteca_df))
biblioteca_df['prezzo'] = rng.uniform(10.0, 50.0, size=len(biblioteca_df)).round(2)

colonne_biblioteca = list(biblioteca_df.columns) # seleziono tutte le colonne
colonne_biblioteca.remove('titolo')
colonne_biblioteca.insert(0,'titolo') # inserisco la colonna titolo come prima colonna
biblioteca_df = biblioteca_df[colonne_biblioteca] # rimpiazzo le colonne vecchie con l'ordine nuovo


print("LISTA LIBRI DELLA BIBLIOTECA\n", biblioteca_df)

# Selezionare libri con più di 300 pagine e creare copia indipendente. 
biblioteca_df_subset = biblioteca_df[biblioteca_df['pagine']>300].copy()

print("\nLISTA LIBRI CON PIU DI 300 PAGINE\n", biblioteca_df_subset)

# Aggiungere colonna “Lungo” (True se pagine > 400). 
biblioteca_df['lungo'] = biblioteca_df['pagine']>400

# Ordinare per prezzo in ordine crescente in-place.
biblioteca_df.sort_values(by='prezzo', inplace=True, ignore_index=True)

print("LISTA LIBRI RIORDINATA PER PREZZO CRESCENTE\n",biblioteca_df.to_string(index=False))
```

```python
""" 
Esercizio 3
Creare un dataframe con clienti (nome, citta, spesa totale). 
Selezionare clienti con spesa > 1000, creare copia indipendente, 
aggiungere colonna “VIP” (true se spesa > 1500), 
ordinare per spesa decrescente e resettare l’indice.

"""

import numpy as np 
import pandas as pd 
rng = np.random.default_rng(112)

n_clienti = 100
clienti_list = [f'Clienti_{i:04d}'for i in range(1, n_clienti + 1)]
nc= len(clienti_list)
citta = [
    "Roma",
    "Milano",
    "Firenze",
    "Napoli",
    "Torino",
    "Parigi",
    "Londra",
    "New York",
    "Tokyo",
    "Berlino",
    "Barcellona",
    "Dublino",
]

spesa_totale = rng.uniform(100.0, 2000.0, size=nc)

clienti_df = pd.DataFrame({
    'cliente_id':clienti_list,
    'citta': rng.choice(citta, size=n),
    'spesa_totale': spesa_totale-round(2)
})

# Selezionare clienti con spesa > 1000, creare copia indipendente
clienti_df_subset = clienti_df[clienti_df['spesa_totale']>1000].copy()

# aggiungere colonna “VIP” (true se spesa > 1500)
clienti_df['VIP'] = clienti_df['spesa_totale'] > 1500 

# ordinare per spesa decrescente e resettare l’indice.
clienti_df_ordered = clienti_df.sort_values(by='spesa_totale',ascending=False, ignore_index=True).reset_index()

print("CLIENTI\n", clienti_df_ordered.drop(columns=['index']).to_string(index=False))
```
