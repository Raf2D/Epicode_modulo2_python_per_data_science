```python
"""
Esercizio 1 
Crea un multiindex con due livelli (Città, Prodotto) e assegna vendite inventate. Stampa tutte le vendite di Milano.
"""

import pandas as pd 
import numpy as np

# Definiamo le liste che comporranno la nostra gerarchia
citta = ("Roma","Milano")
prodotto = ("Prodotto1","Prodotto2")

coppie = list(zip(citta, prodotto))

my_index = pd.MultiIndex.from_tuples(coppie,names=["Citta","Prodotto"])

np.random.seed(42)

df = pd.DataFrame({"Vendite": np.random.randint(10,100, size=len(my_index))},index=my_index)

print("Vendite di Milano: \n", df.loc["Milano"])

```

```python
"""
Esercizio 2
Crea un multiindex da prodotto cartesiano con 2 città, 3 prodotti, 2 mesi. Genera dati casuali e calcola la somma per città e la media per prodotto.
"""

citta = ["Roma","Milano","Cagliari"]
prodotti =["Prodotto1","Prodotto2","Prodotto3"]
mesi =["Gennaio","Febbraio","Marzo"]

index = pd.MultiIndex.from_product(
    [citta, prodotti, mesi], names=["Citta","Prodotto","Mese"]
)

np.random.seed(123)

df2 = pd.DataFrame({
    "Vendite":np.random.randint(1,100,size=len(index))
},index=index)

print("DataFrame originale\n",df2)

print("\nSomma vendite per città: \n", df2.groupby("Citta").sum())
print("\nMedia vendite per città: \n", df2.groupby("Prodotto").mean())
```

```python
""" 
Esercizio 3
Crea un multiindex con 3 livelli (Città, Negozio, Trimestre). Genera vendite casuali e: estrai tutte le vendite del Trimestre Q2, scambia i livelli Città e Trimestre. Riporta il livello Negozio a colonna, calcola la media per città.
"""

citta = ["Roma","Milano","Trieste","Mantova"]
negozio = ["Negozio1","Negozio2","Negozio3"]
trimestre = ["Q1","Q2","Q3","Q4"]

index = pd.MultiIndex.from_product([citta,negozio,trimestre],names=["Citta","Negozio","Trimestre"])
vendite = np.random.randint(1,100, size=len(index))

df3 = pd.DataFrame({"Vendite": vendite}, index=index)

print("Dataframe originale: \n", df3)

print("Vendite del trimestre Q2: \n", df3.xs("Q2",level="Trimestre"))

df_swapped = df3.swaplevel("Citta","Trimestre").sort_index()
df_reset = df3.reset_index("Negozio")
print("Indice Negozio riorganizzato: \n", df_reset)

print("Media vendite per città: \n", df3.groupby("Citta")["Vendite"].mean())
```

```python
""" 
Esercizio 4
Crea un multiindex con 4 livelli (Regione, Città, Prodotto, Mese). Genera dati casuali e poi:
•	Estrai solo i dati della Regione Nord,
•	Calcola la somma per ogni mese, 
•	Ordina l’indice per prodotto,
•	Riportane due livelli a colonna,
•	Trova il prodotto più venduto in ogni città.

"""
regione = ["Regione Nord","Regione Centro","Regione Sud"]
citta = ["Roma","Milano","Cagliari"]
prodotto =["Prodotto1","Prodotto2","Prodotto3"]
mese =["Gennaio","Febbraio","Marzo"]

index4 = pd.MultiIndex.from_product(
    [regione,citta,prodotto,mese],
    names=["Regione","Citta","Prodotto","Mese"]
)
vendite = np.random.randint(1,1000, size=len(index4))
df4 = pd.DataFrame({"Vendite":vendite},index=index4)

print("DataFrame4 originale: \n", df4.head())


# Estrai solo i dati della Regione Nord

print("Dati solo della regione Nord: \n", df4.loc["Regione Nord"])

# Calcola la somma per ogni mese

somma_mensile = df4.groupby("Mese").sum()
print("\nSomma vendite per ogni mese: \n", somma_mensile)

# Ordina l’indice per prodotto
df4_ordinato = df4.sort_index(level="Prodotto")
print("\nDataFrame ordinato per Prodotto: \n", df4_ordinato.head())


# Riportane due livelli a colonna
df4_resettato = df4.reset_index(["Prodotto","Mese"])
print("\nDataFrame resettato per Prodotto e Mese: \n", df4_resettato.head())

# Trova il prodotto più venduto in ogni città

vendite_citta_prod = df4.groupby(["Citta","Prodotto"])["Vendite"].sum().reset_index()
idx_max = vendite_citta_prod.groupby("Citta")["Vendite"].idxmax()
prodotti_top = vendite_citta_prod.loc[idx_max]
print("\n--- 5. Prodotto più venduto per ogni Città ---")
print(prodotti_top.to_string(index=False))
```
