```python
"""
Crea una figura 3x2 subplot dove:
·La prima riga contiene due grafici principali.
·La seconda riga contiene un grafico verticale a
sinistra che occupa due righe.
·La seconda riga a destra contiene un grafico più
piccolo.
Usare tight_layout() per ottimizzare gli spazi.
"""

import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np

# creazione della figura 
fig = plt.figure(figsize=(12,10))

# definizione di una griglia 3x2 tramite GridSpec
gs = gridspec.GridSpec(3,2,figure=fig)

# 1. Prima riga: due grafici principali (occupano la riga 0, colonne 0 e 1)
ax1 = fig.add_subplot(gs[0,0])
ax1.plot(np.linspace(0,10,100), np.sin(np.linspace(0,10,100)), color='tab:blue')
ax1.set_title("Grafico Principale 1")

ax2 = fig.add_subplot(gs[0,1])
ax2.plot(np.linspace(0,10,100), np.cos(np.linspace(0,20,100)), color='tab:orange')
ax2.set_title("Grafico Principale 2")

# 2. Seconda e terza riga a sinistra: grafico verticale che occupa due righe (colonna 0, righe 1 e 2)
ax3 = fig.add_subplot(gs[1:,0])
ax3.plot(np.sin(np.linspace(0,20,100)), np.linspace(0,20,100), color='tab:green')
ax3.set_title("Grafico Verticale (2 righe)")

# 3. Seconda riga a destra: grafico più piccolo (riga 1, colonna 1)
ax4= fig.add_subplot(gs[1,1])
ax4.bar(['A','B','C'],[3,7,5], color='tab:red')
ax4.set_title("Grafico Piccolo Superiore")

# 4. Terza riga e destra: ulteriore grafico per completare la griglia 3x2 (riga 2, colonna 1)
ax5 = fig.add_subplot(gs[2,1])
ax5.scatter(np.random.rand(20), np.random.rand(20), color='tab:purple')
ax5.set_title("Grafico Piccolo Inferiore")

plt.tight_layout()
plt.show()
```
