```python
""" 
Esercizio 1
Creare un DataFrame con 500.000 righe di dati fittizi (vendite, prodotto, regione, prezzo). 
Salvare in CSV, Parquet, Feather e HDF5. 
Misurare i tempi di lettura per ciascun formato e confrontarli.

"""
import numpy as np 
import pandas as pd 
from pathlib import Path 
import time

try:
    percorso_cartella = Path(__file__).resolve().parent
except NameError:
    percorso_cartella = Path.cwd()

n_rows = 500_000
rng = np.random.default_rng(seed=111)

# Liste di categorie da cui estrarre i dati fittizi
prodotti = ["Laptop", "Smartphone", "Tablet", "Monitor", "Cuffie", "Tastiera"]
regioni = ["Nord", "Centro", "Sud", "Isole"]

dati = {
    'prodotto': rng.choice(prodotti, size=n_rows),
    'regione': rng.choice(regioni, size=n_rows),
    'prezzo': rng.uniform(20.0, 2000.0, size=n_rows).round(2),
    'vendite': rng.integers(1, 20, size=n_rows)
}

report_vendite_df = pd.DataFrame(dati)
print(report_vendite_df.head())


# Salvare in CSV, Parquet, Feather e HDF5.

# CSV 
report_vendite_df.to_csv(percorso_cartella/'report_vendite.csv', index=False)
# Parquet 
report_vendite_df.to_parquet(percorso_cartella/'report_vendite.parquet', engine='pyarrow')
# Feather 
report_vendite_df.to_feather(percorso_cartella/'report_vendite.feather')
# HDF5
report_vendite_df.to_hdf(percorso_cartella/'report_vendite.h5', key='vendite', mode='w')

# lettura e tempistiche CSV 
start_date = time.time()
report_vendite_csv = pd.read_csv(percorso_cartella/'report_vendite.csv')
end_date = time.time()
tempo_lettura_csv = end_date - start_date

# lettura e tempistiche PARQUET
start_date = time.time()
report_vendite_parquet = pd.read_parquet(percorso_cartella/'report_vendite.parquet', engine='pyarrow')
end_date = time.time()
tempo_lettura_parquet = end_date - start_date

# lettura e tempistiche FEATHER
start_date = time.time()
report_vendite_feather = pd.read_feather(percorso_cartella/'report_vendite.feather')
end_date = time.time()
tempo_lettura_feather = end_date - start_date

# lettura e tempistiche HDF5
start_date = time.time()
report_vendite_hdf5 = pd.read_hdf(percorso_cartella/'report_vendite.h5', key='vendite')
end_date = time.time()
tempo_lettura_h5 = end_date - start_date

# STAMPA REPORT TEMPI
print("TEMPI DI LETTURA (su 500.000 righe):\n")
print(f"TEMPO CSV:     {tempo_lettura_csv:.4f} secondi")
print(f"TEMPO PARQUET: {tempo_lettura_parquet:.4f} secondi")
print(f"TEMPO FEATHER: {tempo_lettura_feather:.4f} secondi")
print(f"TEMPO HDF5:    {tempo_lettura_h5:.4f} secondi")
```

```python
""" 
Esercizio 3
Salvare due o piu subset diversi nello stesso file HDF5 usando chiavi diverse. 
Leggere solo un subset e calcolare la media del prezzo, 
poi leggere solo le prime 50 righe dello stesso subset usando slicing diretto.

"""

import numpy as np 
import pandas as pd 
from pathlib import Path 
import time

try:
    percorso_cartella = Path(__file__).resolve().parent
except NameError:
    percorso_cartella = Path.cwd()

n_rows = 500_000
rng = np.random.default_rng(seed=111)

# Liste di categorie da cui estrarre i dati fittizi
prodotti = ["Laptop", "Smartphone", "Tablet", "Monitor", "Cuffie", "Tastiera"]
regioni = ["Nord", "Centro", "Sud", "Isole"]

dati = {
    'prodotto': rng.choice(prodotti, size=n_rows),
    'regione': rng.choice(regioni, size=n_rows),
    'prezzo': rng.uniform(20.0, 2000.0, size=n_rows).round(2),
    'vendite': rng.integers(1, 20, size=n_rows)
}

report_vendite_df = pd.DataFrame(dati)
print(report_vendite_df.head())

# Salvare due o piu subset diversi nello stesso file HDF5 usando chiavi diverse. 
report_sub_hdf5_1 = report_vendite_df[report_vendite_df['prezzo'] > 1000]
report_sub_hdf5_2 = report_vendite_df[report_vendite_df['vendite'] > 10]
report_sub_hdf5_3 = report_vendite_df[report_vendite_df['prodotto'] == 'Laptop']

report_sub_hdf5_1.to_hdf(
    percorso_cartella/'report_sub.h5',
    key='report_prezzo',
    format='table', # Fondamentale per il filtraggio
    data_columns=True # Rende tutte le colonne (o alcune specifiche) interrogabili
)
report_sub_hdf5_2.to_hdf(
    percorso_cartella/'report_sub.h5', 
    key='report_vendite',
    format='table',
    data_columns=True
)
report_sub_hdf5_3.to_hdf(
    percorso_cartella/'report_sub.h5',
    key='report_prodotto',
    format='table',
    data_columns=True
)

# Leggere solo un subset e calcolare la media del prezzo
report_sub_hdf5_1 = pd.read_hdf(percorso_cartella/'report_sub.h5', key='report_prezzo')
media_prezzo = report_sub_hdf5_1['prezzo'].mean()
print(f"\nMedia prezzo (subset completo): {media_prezzo:.2f}\n")

# poi leggere solo le prime 50 righe dello stesso subset usando slicing diretto.
report_sub_1_tagliato = report_sub_hdf5_1.iloc[0:50,:]
print("DIMENSIONE SUBSET\n", len(report_sub_1_tagliato))
print("\nPRIME 50 RIGHE DEL SUBSET\n", report_sub_1_tagliato.reset_index())
```

```python
""" 
Esercizio 4
Creare un subset dei dati con vendite elevate e salvarlo in Feather. 
Leggerlo più volte in loop e misurare il tempo totale, 
per dimostrare l’efficienza nei trasferimenti tra processi o calcoli ripetuti

"""

import time
import pandas as pd
import numpy as np
from pathlib import Path

# ==========================================
# 1. GENERAZIONE DATI DI ESEMPIO CON default_rng
# ==========================================
# Inizializziamo il generatore moderno (PCG64) con un seed fisso per la riproducibilità
rng = np.random.default_rng(seed=42)
n_righe = 1_000_000

print("Generazione del dataset iniziale...")
df_totale = pd.DataFrame({
    'id_transazione': f'TRANSAZIONE_{range(1, n_righe + 1)}',
    # Con il generatore moderno usiamo .choice, .uniform e .integers passando il parametro size
    'categoria': rng.choice(['Elettronica', 'Abbigliamento', 'Casa', 'Sport'], size=n_righe),
    'sales': rng.uniform(10, 1000, size=n_righe),
    'quantita': rng.integers(1, 15, size=n_righe) # integers sostituisce randint nel nuovo API
})

# ==========================================
# 2. CREAZIONE DEL SUBSET CON VENDITE ELEVATE
# ==========================================
# Filtriamo i dati per estrarre solo le transazioni con vendite elevate (es. > 900)
soglia = 900
df_subset = df_totale[df_totale['sales'] > soglia].copy()

print(f"Dataset originale: {len(df_totale)} righe")
print(f"Subset con vendite elevate (> {soglia}): {len(df_subset)} righe")

# Percorso del file Feather
try:
    percorso_cartella=Path(__file__).resolve().parent
except:
    percorso_cartella=Path.cwd()

file_path = percorso_cartella/'vendite_elevate.feather'

# ==========================================
# 3. SALVATAGGIO IN FORMATO FEATHER
# ==========================================
# Salviamo il subset in formato binario colonnare per garantire performance elevate 
# nei successivi caricamenti o passaggi tra processi.
df_subset.to_feather(file_path)
print(f"File salvato correttamente in: {file_path}")

# ==========================================
# 4. MISURAZIONE DEL TEMPO DI LETTURA IN LOOP
# ==========================================
# Eseguiamo letture multiple per simulare accessi ripetuti o calcoli iterativi
n_iterazioni = 50
print(f"\nAvvio del test di lettura ripetuta: {n_iterazioni} iterazioni")

tempo_inizio = time.perf_counter()

for _ in range(n_iterazioni):
    df_letto = pd.read_feather(file_path)

tempo_fine = time.perf_counter()
tempo_totale = tempo_fine - tempo_inizio

# ==========================================
# 5. REPORT DEI RISULTATI
# ==========================================
print("-" * 40)
print(f"Tempo totale impiegato: {tempo_totale:.4f} secondi")
print(f"Tempo medio per singola lettura: {tempo_totale / n_iterazioni:.6f} secondi")
print("-" * 40)
```
