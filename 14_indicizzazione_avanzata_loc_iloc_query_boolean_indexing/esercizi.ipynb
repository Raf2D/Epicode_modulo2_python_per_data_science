```python
""" 
Esercizio 1
-   Crea un DataFrame con colonne: Studente, Materia, Voto, Classe, Eta.
-   Seleziona tutti gli studenti con voto maggiore di 28, in “matematica” usando loc e boolean indexing.
-   Ordina i risultati per voto decrescente.
-   Aggiungi una colonna che indica “ottimo” se voto >= 30, “buono” se voto >= 28 e “sufficiente” altrimenti.
-   Calcola la media dei voti e crea una colonna booleana “sopra media”.
"""

import numpy as np 
import pandas as pd
from pathlib import Path

# Gestione percorso
try:
    percorso_cartella = Path(__file__).parent.resolve()
except NameError:
    percorso_cartella = Path.cwd()

n_rows = 100

# GENERATORE UNICO CON SEED PER LA RIPRODUCIBILITÀ
rng = np.random.default_rng(seed=123)

studenti_ids = [f"STUDENTE_{i:04d}" for i in range(1, n_rows + 1)]
materie = ["Informatica", "Lettere latine", "Geometria", "Analisi matematica", "Fisica generale"]
classi = ["A1", "A2", "B1", "B2"]

# Date casuali
start_date = np.datetime64('2024-01-01')
end_date = np.datetime64('2026-12-31')
giorni_totali = (end_date - start_date).astype(int)
giorni_casuali = rng.integers(0, giorni_totali, size=n_rows)
date_casuali = start_date + giorni_casuali

# Creazione DataFrame (tutti usano lo stesso rng)
studenti = pd.DataFrame({
    "Studente": rng.choice(studenti_ids, size=n_rows),
    "Materia": rng.choice(materie, size=n_rows),
    "Voto": rng.uniform(18, 31, size=n_rows).round(2),
    "Classe": rng.choice(classi, size=n_rows),
    "Eta": rng.integers(18, 35, size=n_rows, endpoint=True),
    "Data": date_casuali
})


# Seleziona studenti con voto > 28 in "Analisi matematica"
studenti_subset = studenti.loc[(studenti['Voto'] > 28) & (studenti['Materia'] == 'Analisi matematica')].copy()

# Ordina i risultati per voto decrescente
studenti_subset.sort_values(by='Voto', ascending=False, inplace=True)
print(f'\nTutti i {len(studenti_subset)} studenti con voto maggiore di 28 in Analisi matematica:\n', studenti_subset)

# Aggiungi una colonna che indica “ottimo” se voto >= 30, “buono” se voto >= 28 e “sufficiente” altrimenti.
studenti['Esito'] = studenti['Voto'].apply(lambda x: 'ottimo' if x >= 30 else ('buono' if x >= 28 else 'sufficiente'))

print("\nTutti gli studenti\n", studenti)

# Calcola la media dei voti e crea una colonna booleana “sopra media”.
studenti_media_voto = studenti['Voto'].mean()

studenti['Sopra media'] = studenti['Voto'].apply(lambda x: True if x > studenti_media_voto else False).copy()

# Salva CSV
studenti.to_csv(percorso_cartella / 'studenti.csv', index=False)
print("Dataframe STUDENTI (prime 5 righe):\n", studenti.sort_values(by='Voto', ascending=False).copy().head())

```

```python
""" 
Esercizio 2
•	Crea un dataframe con colonne: Prodotto, categoria, prezzo, quantità, disponibilità.
•	Usa query per selezionare i prodotti della categoria “tecnologia” con prezzo tra 100 e 300.
•	Dal risultato, usa iloc per prendere le prime tre righe e solo le colonne prodotto e prezzo.
•	Aggiungi una colonna che calcola il fatturato: prezzo * quantità
•	Ordina il DataFrame per fatturato decrescente e calcola la media del fatturato dei prodotti selezionati.

"""

import numpy as np 
import pandas as pd 
from pathlib import Path
import random

n_id_prodotti = 100
n_rows = 10

try: 
    percorso_cartella = Path(__file__).resolve().parent
except NameError:
    percorso_cartella = Path.cwd()

# GENERATORE UNICO CON SEED PER LA RIPRODUCIBILITÀ
rng = np.random.default_rng(seed=123)

prodotti_ids = [f"PROD_ {i:04d}" for i in range(n_id_prodotti + 1)] # lista prodotti unici
n = len(prodotti_ids)
quantita_prodotti = np.random.randint(0,10,size=n)

categorie_prodotti = [
    "Elettronica",
    "Abbigliamento",
    "Casa e Cucina",
    "Sport e Tempo Libero",
    "Libri e Cancelleria",
    "Salute e Bellezza",
    "Alimentari e Bevande",
    "Giocattoli e Infanzia",
    "Tecnologia",
]

catalogo_list = {
        "id": prodotti_ids,
        "prezzo_unitario": np.round(rng.uniform(10.0, 500.0, size=n),2),
        "categoria": rng.choice(categorie_prodotti,size=n),
        "disponibilita": quantita_prodotti > 0,
        "quantita": quantita_prodotti
    }

catalogo_df = pd.DataFrame(catalogo_list)

print("CATALOGO:\n", catalogo_df)

# Usa query per selezionare i prodotti della categoria “tecnologia” con prezzo tra 100 e 300.
catalogo_subset = catalogo_df.query("(prezzo_unitario >= 100 and prezzo_unitario <= 300) and categoria=='Tecnologia'").copy().reset_index()
print("\nPRODOTTI CON PREZZO TRA 100 e 300\n", catalogo_subset)


# Dal risultato, usa iloc per prendere le prime tre righe e solo le colonne prodotto e prezzo.
print("\n3 PRODOTTI - Tecnologia con solo ID e PREZZO :\n",catalogo_subset.iloc[0:3,1:3])

# Aggiungi una colonna che calcola il fatturato: prezzo * quantità
catalogo_df['fatturato'] = catalogo_df["prezzo_unitario"] * catalogo_df["quantita"]

print("\nNUOVA COLONNA FATTURATO AGGIUNTA\n",catalogo_df)

# Ordina il DataFrame per fatturato decrescente e calcola la media del fatturato dei prodotti selezionati.
catalogo_df_ordered = catalogo_df.copy().sort_values(by='fatturato', ascending=False).reset_index()
media_prezzo_prodotti = catalogo_df_ordered.query("categoria=='Elettronica'")['prezzo_unitario'].mean()
print(f"\nMEDIA DEL PREZZO RIGUARDO PRODOTTI DI CATEGORIA Elettronica\n {media_prezzo_prodotti:.2f}€")
```

```python
""" 
Esercizio 3
•	Crea un dataframe con colonne: dipendente, reparto, eta, stipendio, anniservizio
•	Usa boolean indexing per selezionare i dipendenti con eta>30 e stipendio>2000, oppure appartenenti al reparto “IT”.
•	Mostra solo dipendente, eta e stipendio.
•	Aggiungi una colonna che indica “Junior” se eta < 35 e “Senior” altrimenti
•	Raggruppa per reparto e calcola la media dello stipendio per reparto e per categoria.
•	Ordina i risultati per stipendio medio decrescente.

"""

import numpy as np
import pandas as pd 
from pathlib import Path 

try:
    percorso_cartella = Path(__file__).resolve().parent
except NameError:
    percorso_cartella = Path.cwd()

n_dipendenti = 100
rng = np.random.default_rng(seed = 21)
dipendenti_list = [f"Dipendente_{i:04d}" for i in range(1, n_dipendenti + 1)]
reparto_list = [
    "Risorse Umane (HR)",
    "Amministrazione e Finanza",
    "Marketing e Comunicazione",
    "Vendite e Commerciale",
    "Information Technology (IT)",
    "Ricerca e Sviluppo (R&D)",
    "Produzione e Operazioni",
    "Logistica e Supply Chain",
    "Servizio Clienti (Customer Care)",
    "Legale e Compliance",
    "Acquisti e Procurement",
    "Controllo di Gestione"
]

anni_servizio = rng.integers(1, 20, size=n_dipendenti)
stipendi = rng.uniform(1500.0, 10_000.0, size=n_dipendenti)
eta = rng.integers(18, 67, size=n_dipendenti)


dipendenti_df = pd.DataFrame({
    "dipendente_id":dipendenti_list,
    "reparto": rng.choice(reparto_list, size=n_dipendenti),
    "eta": eta,
    "stipendio": stipendi.round(2),
    "anni_servizio": anni_servizio,
})

print("LISTA DIPENDENTI:\n", dipendenti_df)

# Usa boolean indexing per selezionare i dipendenti con eta>30 e stipendio>2000, oppure appartenenti al reparto “IT”.
# Mostra solo dipendente, eta e stipendio.
dipendenti_df_subset = dipendenti_df.loc[
    ((dipendenti_df['eta']>30) & (dipendenti_df['stipendio']>2000)) | 
    (dipendenti_df['reparto']=='Information Technology (IT)'),
    ['dipendente_id','eta','stipendio']
].copy()

print("\nLISTA DIPENDENTI CON ETA > 30  E STIPENDIO > 2000 oppure APPARTENENTI AL REPARTO IT\n",dipendenti_df_subset)

# Aggiungi una colonna che indica “Junior” se eta < 35 e “Senior” altrimenti
dipendenti_df['categoria'] = dipendenti_df['eta'].apply(lambda x: 'junior' if x<35 else 'senior')
print("\nLISTA DIPENDENTI:\n", dipendenti_df)

# Raggruppa per reparto e calcola la media dello stipendio per reparto e per categoria.
media_stipendio_reparto_categ = dipendenti_df.groupby(['reparto','categoria'])['stipendio'].mean()

#Ordina i risultati per stipendio medio decrescente.
media_stipendio_reparto_categ.sort_values(ascending=False)
print("\nMEDIA STIPENDI PER REPARTO E PER CATEGORIA:\n", np.round(media_stipendio_reparto_categ.sort_values(ascending=False),2))
```
